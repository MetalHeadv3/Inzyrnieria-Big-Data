```python
from pyspark.sql import SparkSession
import requests
import string
import json

spark = SparkSession.builder.appName("PanTadeuszBagOfWords").getOrCreate()
```

```python
rdd = spark.sparkContext.textFile("pan-tadeusz.txt")
```

```python
stopwords_url = "https://raw.githubusercontent.com/bieli/stopwords/master/polish.stopwords.txt"
stopwords = set(requests.get(stopwords_url).text.splitlines())
```

```python
def clean_word(word):
    return word.strip(string.punctuation).lower()

words = rdd.flatMap(lambda line: line.split()) \
           .map(clean_word) \
           .filter(lambda word: word and word not in stopwords)
```

```python
word_counts = words.map(lambda word: (word, 1)).reduceByKey(lambda a, b: a + b)
word_dict = dict(word_counts.collect())

```

```python
with open("pan_tadeusz_bag_of_words.json", "w", encoding="utf-8") as f:
    json.dump(word_dict, f, ensure_ascii=False, indent=2)
```

```python
most_common = max(word_dict.items(), key=lambda x: x[1])
print("Najczęściej występujące słowo:", most_common[0])
print("Liczba wystąpień:", most_common[1])
```
