```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("RecipeReviewsAnalysis").getOrCreate()

df = spark.read.csv("Recipe Reviews and User Feedback Dataset.csv", header=True, inferSchema=True)
df.show(5)

```

```python
df = df.withColumnRenamed("_c0", "id")
df.printSchema()
```

```python
df.select("reply_count").orderBy("reply_count", ascending=False).show(10)
```

```python
df.groupBy("recipe_code") \
  .sum("best_score") \
  .orderBy("sum(best_score)", ascending=False) \
  .show(10)

```

```python
df.groupBy("recipe_code") \
  .count() \
  .orderBy("count", ascending=False) \
  .show(10)
```

```python
df.groupBy("stars").count().orderBy("stars").show()
```
