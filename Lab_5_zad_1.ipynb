```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("PanTadeuszAnalysis").getOrCreate()
rdd = spark.sparkContext.textFile("pan-tadeusz.txt")  # podaj pełną ścieżkę jeśli plik nie jest w katalogu roboczym
```

```python
rdd.count()
```

```python
rdd.map(lambda line: len(line)).top(10)
```

```python
rdd.sortBy(lambda line: len(line), ascending=False).take(10)
```

```python
words = rdd.flatMap(lambda line: line.split())
unique_words = words.distinct().collect()
len(unique_words)
```

```python
word_counts = words.countByValue()
list(word_counts.items())[:10]
```

```python
sorted(word_counts.items(), key=lambda x: x[1], reverse=True)[:10]
```

```python
rdd.filter(lambda line: len(line.split()) > 7).take(10)
```

```python
rdd.filter(lambda line: "Tadeusz" in line).collect()
```

```python
[last.lower() for last in rdd.collect()[-10:]]
```

```python
rdd.map(lambda line: len(line)).reduce(lambda a, b: a + b)
```
